### Python in 10 minutes (and how it differs from R)

Python is the language that the code cells understand. If you already know R, most of the ideas are the same, but some details are different and they are a classic source of bugs. Here is a short translation table that you can come back to:

| | R | Python |
| --- | --- | --- |
| assign | `x <- 3` | `x = 3` |
| first element | `x[1]` | `x[0]` &nbsp; (**Python counts from 0**) |
| a range | `1:6` &nbsp; (includes 6) | `range(1, 7)` or `np.arange(1, 7)` &nbsp; (**the end, 7, is excluded**) |
| a vector | `c(1, 2, 3)` | `np.array([1, 2, 3])` &nbsp; (or a *list* `[1, 2, 3]`) |
| standard deviation | `sd(x)` &nbsp; (divides by N−1) | `np.std(x, ddof=1)` &nbsp; (`np.std(x)` divides by N) |
| random numbers | `rnorm(5)`, `sample(1:6, 5, replace=TRUE)` | `rng.normal(size=5)`, `rng.integers(1, 7, size=5)` |
| use a package | `library(ggplot2)` | `import matplotlib.pyplot as plt`, then `plt.plot(...)` |
| a block of code | `{ ... }` | **indentation** (the lines after a `:` that are shifted to the right) |
| a named list | `list(a = 1)` | a *dictionary* `{'a': 1}` |
| plots | `plot()`, `hist()`, ggplot2 | matplotlib: `plt.plot()`, `plt.hist()` |

### The PDB format

The coordinates are in the lines that start with `ATOM` (and `HETATM` for ligands, ions, water, ...). A PDB file is a **fixed-width** format: every piece of information is in precise columns of the line. We read the file in Python as a list of lines, we take the `ATOM` lines, and we look at the first one using slices (remember: the first column is number 0 and the end of a slice is excluded).

```
with open('7tjh.txt') as f:                        # open the file for reading
    lines = [l.rstrip('\n') for l in f]            # a list with all the lines, without the end-of-line character
print(len(lines), 'lines')

atom_lines = [l for l in lines if l.startswith('ATOM')]       # keep only the ATOM lines
print(len(atom_lines), 'ATOM lines')

l = atom_lines[0]
print(l)
print('record name  :', repr(l[0:6]))
print('atom name    :', repr(l[12:16]))     # columns 13-16
print('alt. location:', repr(l[16]))        # column 17, blank or A, B, ... if the atom has alternative positions
print('residue name :', repr(l[17:20]))     # columns 18-20
print('chain        :', repr(l[21]))        # column 22
print('residue num. :', repr(l[22:26]))     # columns 23-26
print('x, y, z      :', float(l[30:38]), float(l[38:46]), float(l[46:54]))   # columns 31-54
```

### Counting amino acids, first attempt: the shell

```
%%bash
# show the first lines of the file that include ALA (ALANINE atoms)
grep ALA 7tjh.txt | head -n 10
```

`grep` is quick and handy, but it is also imprecise: it finds the text *anywhere* in the line, not in the right column. So for example `CA` is also the symbol of calcium, and a residue with two alternative positions is counted twice. For a precise count it is better to read the right columns, as we did above.

### Counting amino acids, second attempt: Python

Every amino acid has exactly one `CA` atom (the alpha carbon). So counting the `CA` atoms of the `ATOM` lines, and reading their residue name, gives the amino acid composition. If an atom has alternative positions (column 17 is not blank) we count only the first one. A **`Counter`** is a dictionary specialised in counting: `Counter(list)` returns, for every different element of the list, how many times it appears (and 0 for the elements that are not there).

```
from collections import Counter

# the residue name of every CA atom (atom name ' CA ') that is in the main position (blank or 'A')
residues = [l[17:20] for l in atom_lines if l[12:16] == ' CA ' and l[16] in ' A']
print(len(residues), 'amino acids in total')

aa_count = Counter(residues)
print(aa_count)
```

### 1D discrete data: throwing a die

A die is the simplest stochastic process: the outcome is one of 6 possibilities, and it is random. Let's begin by throwing it once: this means to generate an integer number between 1 and 6 extracted from a uniform probability distribution. Run the cell several times, you will always get a different number.

```
rng.integers(low=1, high=7)     # here the range is from low to high-1
```

Now we decide how many times to throw it and we collect the outcomes. This is **sampling**: exploring a probability distribution by looking at many outcomes.

```
num_throw = 10000
data = rng.integers(low=1, high=7, size=num_throw)
print(data[:20], '...', len(data), 'outcomes')
```

**Histograms.** To estimate the probabilities we count how many times each face appears (a histogram) and we make a bar plot.

```
faces = np.arange(1, 7)                          # 1, 2, ..., 6
counts = np.bincount(data, minlength=7)[1:]      # counts[0] is how many times we got 1, ..., counts[5] how many times 6
print(faces)
print(counts)

plt.bar(faces, counts)
plt.xlabel('face')
plt.ylabel('counts')
plt.title('Histogram of a die throwing experiment')
plt.show()
```

**Absolute and relative probabilities.** As in the lecture:

* *absolute probability*: $p(i) = \mathrm{count}(i) / \sum_k \mathrm{count}(k)$. To compute it we need to know **all** the outcomes, it is normalised: $\sum_i p(i)=1$;
* *relative probability*: $p(j)/p(i) = \mathrm{count}(j)/\mathrm{count}(i)$. It only needs the two counts, it is just a frequency ratio.

```
prob = counts / np.sum(counts)                  # absolute probabilities
print('absolute probabilities:', np.round(prob, 4), ' sum =', np.sum(prob))
print('relative probability p(2)/p(1) from the counts:', counts[1]/counts[0])
print('relative probability p(2)/p(1) from the probabilities:', prob[1]/prob[0], ' (expected 1 for a fair die)')

fig, axes = plt.subplots(1, 2, figsize=(9, 3.5))
axes[0].bar(faces, prob, width=0.8)
axes[0].set_title('bar plot of the probabilities')
# plt.hist does the counting for us; density=True normalises the histogram
axes[1].hist(data, bins=np.arange(0.5, 7.5, 1), density=True, rwidth=0.8)
axes[1].set_title('plt.hist(data, density=True)')
for ax in axes:
    ax.axhline(1/6, color='k', ls='--', label='expected 1/6')   # horizontal line
    ax.set_xlabel('face')
    ax.set_ylabel('probability')
    ax.legend()
plt.tight_layout()
plt.show()
```

**The problem of sampling.** The histogram is never perfectly flat: with a finite number of throws the estimated probabilities fluctuate around the true value. Let's see what happens as the number of throws increases.

```
fig, axes = plt.subplots(1, 4, figsize=(14, 3), sharey=True)
for ax, n in zip(axes, [10, 100, 1000, 100000]):
    d = rng.integers(low=1, high=7, size=n)
    p = np.bincount(d, minlength=7)[1:] / n
    ax.bar(faces, p)
    ax.axhline(1/6, color='k', ls='--')
    ax.set_title(f'{n} throws')
    ax.set_xlabel('face')
axes[0].set_ylabel('probability')
plt.tight_layout()
plt.show()

# the typical size of the fluctuation of the estimated probability p is sqrt(p(1-p)/N)
p = 1/6
for n in [10, 100, 1000, 10000, 100000]:
    fluct = np.sqrt(p*(1-p)/n)
    print(f'{n:>7d} throws: typical error on p = {fluct:.4f}  ({100*fluct/p:.1f}% of 1/6)')
```

The uniform distribution is not special: it needs the same number of throws as any other. The more precise we want the probabilities, the more outcomes we need: the error decreases as $1/\sqrt{N}$.

**Average, standard deviation and standard error.** Experiments rarely show single outcomes, but averages. Using the definitions of the lecture:

$$\langle x \rangle = \frac{1}{N}\sum_{i=1}^N x_i \qquad \sigma = \sqrt{\frac{1}{N}\sum_{i=1}^N (x_i-\langle x \rangle)^2} \qquad \text{std-err} = \frac{\sigma}{\sqrt{N}} = \sqrt{\frac{1}{N^2-N}\sum_{i=1}^N (x_i-\langle x \rangle)^2}$$

The standard deviation $\sigma$ is the *width of the distribution*; the standard error is the *precision of our estimate of the average*.

```
N = len(data)
average = np.sum(data)/N
deviation = np.sqrt(np.sum((data-average)**2)/N)
error = np.sqrt(np.sum((data-average)**2)/(N**2-N))

print('by hand :', average, deviation, error)
# the same with the functions of numpy and scipy
print('numpy   :', np.average(data), np.std(data), sem(data))
print('exact   : 3.5', np.sqrt(35/12))

# the error tells us how many digits are meaningful: as a rule of thumb do not use more than
# two digits starting from the first non-zero one of the error
print(f'average = {average:.3f} +/- {error:.3f}')
```

What changes when we increase the number of throws? The **standard deviation** stays the same (it is a property of the die), while the **standard error** goes down as $1/\sqrt{N}$ (it is a property of our estimate of the average).

```
Ns = [10, 100, 1000, 10000, 100000, 1000000]
avs, devs, errs = [], [], []
for n in Ns:
    d = rng.integers(low=1, high=7, size=n)
    avs.append(np.average(d))
    devs.append(np.std(d))
    errs.append(sem(d))
    print(f'N = {n:>8d}   average = {avs[-1]:.3f}   std = {devs[-1]:.3f}   std-err = {errs[-1]:.4f}')

plt.errorbar(Ns, avs, yerr=errs, fmt='o', capsize=4, label='average +/- std-err')
plt.axhline(3.5, color='k', ls='--', label='exact: 3.5')
plt.xscale('log')                        # logarithmic axis
plt.xlabel('number of throws N')
plt.ylabel('average')
plt.legend()
plt.show()
```

**The average of many throws is Gaussian.** What often happens in an experiment is that we observe a number that is already an average: for example multiple copies ($n$) of the same protein give a fluorescent signal between 1 and 6 ;-) and we measure the average. Here this is the average over $n$ throws, and we repeat the experiment many times (`num_rep`). What is the probability distribution of these averages? According to the lecture (central limit theorem) it is a Gaussian centred on the average and with a width equal to the standard error $\sigma/\sqrt{n}$.

```
num_rep = 2000                      # number of replicated experiments
sigma_die = np.sqrt(35/12)          # standard deviation of one throw

fig, axes = plt.subplots(1, 4, figsize=(14, 3))
for ax, n in zip(axes, [1, 2, 10, 100]):
    # every row is one experiment of n throws, the outcome of the experiment is the average over the row
    adata = rng.integers(low=1, high=7, size=(num_rep, n)).mean(axis=1)
    err = sigma_die/np.sqrt(n)                        # expected standard error
    # the possible averages are multiples of 1/n: we choose the bins accordingly
    step = 1/n
    width = step*max(1, round(err/(4*step)))
    edges = np.arange(1-step/2, 6+step/2+width, width)
    ax.hist(adata, bins=edges, density=True, color='lightgrey', edgecolor='grey')
    xg = np.linspace(1, 6, 400)
    ax.plot(xg, norm.pdf(xg, 3.5, err), 'k', label='Gaussian')
    ax.set_title(f'average of {n} throws')
    ax.set_xlabel('average')
    print(f'n = {n:>3d}: std of the averages = {np.std(adata):.3f},  expected std-err = {err:.3f}')
axes[0].set_ylabel('probability density')
axes[0].legend()
plt.tight_layout()
plt.show()
```

### A biased die

If we add a small weight inside the die we change its mechanics (the centre of mass) in the presence of gravity, and one face becomes more likely. Let's say that face 1 has probability $p(1)=0.2$ and all the others $p=0.16$ (as in the lecture). We can ask `rng.choice` to extract the faces with given probabilities.

```
p_biased = np.array([0.2, 0.16, 0.16, 0.16, 0.16, 0.16])
print('normalised?', np.sum(p_biased))

data_b = rng.choice(faces, size=10000, p=p_biased)
prob_b = np.bincount(data_b, minlength=7)[1:] / len(data_b)

plt.bar(faces, prob_b, label='estimated from 10000 throws')
plt.plot(faces, p_biased, 'ko', label='exact')
plt.xlabel('face')
plt.ylabel('probability')
plt.title('A biased die')
plt.legend()
plt.show()
```

How many throws do we need to *see* the bias? Here every probability is shown with its error $\sqrt{p(1-p)/N}$ (an error bar): the bias is visible only when the error bar of face 1 does not overlap with the others.

```
fig, axes = plt.subplots(1, 4, figsize=(14, 3), sharey=True)
for ax, n in zip(axes, [100, 1000, 10000, 100000]):
    d = rng.choice(faces, size=n, p=p_biased)
    p = np.bincount(d, minlength=7)[1:] / n
    ax.errorbar(faces, p, yerr=np.sqrt(p*(1-p)/n), fmt='o', capsize=3)
    ax.plot(faces, p_biased, 'kx')                # exact values
    ax.set_title(f'{n} throws')
    ax.set_xlabel('face')
axes[0].set_ylabel('probability')
plt.tight_layout()
plt.show()
```

### From energies to probabilities: the Boltzmann distribution

The connection between statistics and mechanics is the **Boltzmann distribution**: a system in thermal contact with a bath at temperature $T$ visits a configuration $x$ with a probability density

$$\rho(x) \propto \exp\left[-\frac{U(x)}{k_BT}\right]$$

If some outcome is more likely than another, we can think that it is *favourable in energy*. For a die:

* a **fair die** has a flat potential energy, all faces have the same $U$ and so the same probability;
* in the **biased die** face 1 has a lower energy. How much lower? From $p(1)/p(2) = \exp[-(U_1-U_2)/k_BT] = 0.2/0.16 = 1.25$ we get $U_1-U_2 = -k_BT\ln 1.25 \simeq -0.223\,k_BT$.

To go from energies to probabilities: compute the Boltzmann weights $\exp(-U/k_BT)$ and **normalise** them dividing by their sum $Z$ (the sum of the weights, also called partition function).

```
kT = 1.0                                   # our unit of energy

def boltzmann(U, kT=1.0):
    weights = np.exp(-U/kT)                # Boltzmann weights (not normalised)
    Z = np.sum(weights)                    # normalisation: sum over all the outcomes
    return weights/Z

U_fair = np.zeros(6)                       # flat potential energy
U_bias = np.zeros(6)
U_bias[0] = -kT*np.log(1.25)               # face 1 is lower in energy by 0.223 kT

print('fair  :', np.round(boltzmann(U_fair, kT), 3))
print('biased:', np.round(boltzmann(U_bias, kT), 3))

fig, axes = plt.subplots(2, 2, figsize=(9, 5), sharex=True)
for j, (U, name) in enumerate([(U_fair, 'fair die'), (U_bias, 'biased die')]):
    axes[0, j].plot(faces, U, 'o-')
    axes[0, j].set_title(name)
    axes[0, j].set_ylabel('energy U (kT)')
    axes[0, j].set_ylim(-0.5, 0.5)
    axes[1, j].bar(faces, boltzmann(U, kT))
    axes[1, j].set_ylabel('probability')
    axes[1, j].set_xlabel('face')
plt.tight_layout()
plt.show()
```

**And back: from probabilities to energies.** This is the key idea of an *in-silico microscope*: if we can sample the probabilities, then we can estimate the energy differences as

$$U_j - U_i = -k_BT \ln\frac{p(j)}{p(i)} = -k_BT \ln\frac{\mathrm{count}(j)}{\mathrm{count}(i)}$$

(a relative probability is enough!). The error on the counts propagates to the energy: $\sigma_{U_j-U_i} \simeq k_BT\sqrt{1/\mathrm{count}(j)+1/\mathrm{count}(i)}$. Let's use face 2 as reference and see how the estimate improves with the number of throws.

```
def energies_from_samples(d, kT=1.0, ref=1):
    # energies of all faces relative to the reference face (ref=1 means face 2) and their errors
    c = np.bincount(d, minlength=7)[1:]
    dU = -kT*np.log(c/c[ref])
    err = kT*np.sqrt(1/c + 1/c[ref])
    err[ref] = 0
    return dU, err

p_bias = boltzmann(U_bias, kT)

fig, axes = plt.subplots(1, 4, figsize=(14, 3), sharey=True)
for ax, n in zip(axes, [100, 1000, 10000, 1000000]):
    d = rng.choice(faces, size=n, p=p_bias)
    dU, err = energies_from_samples(d, kT)
    ax.errorbar(faces, dU, yerr=err, fmt='o', capsize=3, label='estimated')
    ax.plot(faces, U_bias - U_bias[1], 'kx', label='exact')
    ax.set_title(f'{n} throws')
    ax.set_xlabel('face')
axes[0].set_ylabel('U - U(face 2)  (kT)')
axes[0].legend()
plt.tight_layout()
plt.show()
```

### 1D continuous data: from the die to the wheel

What if the outcomes are not 6, but 12, 36, ... or infinite? Think of a **wheel** with a pointer: the outcome is the angle $\theta$ where it stops. A die is a wheel with 6 equal sectors. If we increase the number of sectors, the probability of each sector shrinks ($1/6$, $1/12$, $1/36$, ...), but the **probability per unit angle** (the probability density $\rho = p/\Delta\theta$) stays constant. For the continuous angle the outcome of the experiment is described by a density $\rho(\theta)$, here constant and equal to $1/360$ per degree (or $1/2\pi$ per radian).

```
num_spin = 100000
theta = rng.uniform(low=-180, high=180, size=num_spin)     # angles in degrees, all equally likely

fig, axes = plt.subplots(2, 4, figsize=(14, 5), sharex=True)
for j, nsec in enumerate([6, 12, 36, 360]):
    counts_w, edges = np.histogram(theta, bins=nsec, range=(-180, 180))
    width = edges[1] - edges[0]                  # size of one sector (degrees)
    p_sector = counts_w / num_spin               # probability of one sector
    rho = p_sector / width                       # probability density (1/degree)
    axes[0, j].bar(edges[:-1], p_sector, width=width, align='edge')
    axes[0, j].set_ylim(0, 0.2)
    axes[0, j].set_title(f'{nsec} sectors:  p = {1/nsec:.3f}')
    axes[1, j].bar(edges[:-1], rho, width=width, align='edge')
    axes[1, j].axhline(1/360, color='k', ls='--')
    axes[1, j].set_ylim(0, 2/360)
    axes[1, j].set_xlabel('angle (degrees)')
axes[0, 0].set_ylabel('probability p')
axes[1, 0].set_ylabel('density p/width (1/degree)')
plt.tight_layout()
plt.show()
```

For a continuous variable the probability is not given by the height of the density but by its **integral**: the probability that the angle is between $a$ and $b$ is $P(a<\theta<b)=\int_a^b \rho(\theta)\,d\theta$. For the fair wheel $\rho=1/360$ and so, for example, $P(20^\circ<\theta<50^\circ) = 30/360 \simeq 8.3\%$. We can check it by counting the spins that end in this interval.

Note that **densities have units of 1/x**: if we change the unit of the angle (degrees $\to$ radians) the density changes, the probability (and the integral of the density, 1) does not.

```
a, b = 20, 50
inside = (theta > a) & (theta < b)                 # True/False for each spin
p_count = np.mean(inside)                          # fraction of spins in the interval
p_err = np.sqrt(p_count*(1-p_count)/num_spin)
print(f'from the spins: P({a} < theta < {b}) = {100*p_count:.2f} +/- {100*p_err:.2f} %')
print(f'from the density: rho * (b-a) = {100*(b-a)/360:.2f} %')

# the same data in degrees and in radians
for unit, d, lo, hi in [('degree', theta, -180, 180), ('radian', np.radians(theta), -np.pi, np.pi)]:
    rho, edges = np.histogram(d, bins=36, range=(lo, hi), density=True)
    print(f'density in 1/{unit}: {np.mean(rho):.5f},   integral of the density = {np.sum(rho*np.diff(edges)):.3f}')
```

### The wheel with an added weight

Now we add a weight to the rim of the wheel. As for the biased die, we change the mechanics: in the presence of gravity the energy depends on the angle $\theta$ between the weight and the pointer,

$$U(\theta) = \varepsilon\,(1-\cos\theta)$$

with a minimum at $\theta=0$ (weight at the pointer, $\varepsilon = m g r$ is the energy scale of the weight). The wheel is kicked all the time (by the air, or for a molecule by the solvent) with a thermal energy of the order of $k_BT$, so the angle is a stochastic variable distributed according to Boltzmann:

$$\rho(\theta) = \frac{1}{Z}\exp\left[-\frac{U(\theta)}{k_BT}\right] \qquad Z = \int \exp\left[-\frac{U(\theta)}{k_BT}\right] d\theta$$

where $Z$ makes sure that $\int\rho\,d\theta = 1$. We compute it on a fine grid of angles (the integral becomes a sum times the grid spacing $\Delta\theta$).

```
kT = 1.0
eps = 5.0*kT                                       # the weight is heavy: eps = 5 kT

dtheta = 0.1                                       # grid spacing (degrees)
grid = np.arange(-180, 180, dtheta) + dtheta/2     # centre of each cell of the grid
U = eps*(1 - np.cos(np.radians(grid)))             # potential energy

weights = np.exp(-U/kT)
Z = np.sum(weights)*dtheta                         # integral of the Boltzmann weights
rho_w = weights/Z                                  # normalised probability density (1/degree)
print('integral of rho =', np.sum(rho_w)*dtheta)

fig, axes = plt.subplots(1, 2, figsize=(9, 3.5))
axes[0].plot(grid, U/kT)
axes[0].set_ylabel('energy U (kT)')
axes[1].plot(grid, rho_w)
axes[1].set_ylabel('probability density (1/degree)')
for ax in axes:
    ax.set_xlabel('angle (degrees)')
plt.tight_layout()
plt.show()
```

Now let's do the experiment: we spin the wheel 3000 times (as in the lecture). In a real wheel the thermal kicks do the job; here we draw angles from the grid, each one with a probability proportional to its Boltzmann weight (in the next lectures we will see how to do the same with a simulation of the actual dynamics). We compare the histogram with the Boltzmann density, and we calculate the probability of the interval $20^\circ<\theta<50^\circ$ for the wheel with the weight.

```
num_spin = 3000
theta_w = rng.choice(grid, size=num_spin, p=weights/np.sum(weights))

plt.hist(theta_w, bins=60, range=(-180, 180), density=True, color='lightgrey', label=f'{num_spin} spins')
plt.plot(grid, rho_w, 'k', label='Boltzmann: exp(-U/kT)/Z')
plt.axvspan(20, 50, color='tab:blue', alpha=0.3, label='20 < theta < 50')
plt.xlabel('angle (degrees)')
plt.ylabel('probability density (1/degree)')
plt.title('Wheel with an added weight')
plt.legend(loc='upper left')
plt.show()

inside = (theta_w > 20) & (theta_w < 50)
p_count = np.mean(inside)
p_err = np.sqrt(p_count*(1-p_count)/num_spin)
p_int = np.sum(rho_w[(grid > 20) & (grid < 50)])*dtheta
print(f'from the spins   : P(20 < theta < 50) = {100*p_count:.1f} +/- {100*p_err:.1f} %')
print(f'integral of rho  : P(20 < theta < 50) = {100*p_int:.1f} %      (fair wheel: 8.3 %)')
```

**The Gaussian approximation.** Close to the minimum $\cos\theta \simeq 1-\theta^2/2$, so $U\simeq \varepsilon\theta^2/2$ ($\theta$ in radians): a *harmonic* energy gives a **Gaussian** distribution with $\langle\theta\rangle=0$ and $\sigma^2 = k_BT/\varepsilon$,

$$\rho_G(\theta)=\frac{1}{\sqrt{2\pi}\sigma}\exp\left[-\frac{(\theta-\langle\theta\rangle)^2}{2\sigma^2}\right]$$

We can now use average, standard deviation and standard error to describe the spins.

```
sigma_deg = np.degrees(np.sqrt(kT/eps))            # width of the Gaussian approximation (degrees)

average = np.average(theta_w)
deviation = np.std(theta_w)
error = sem(theta_w)
print(f'average   = {average:.1f} +/- {error:.1f} degrees  (expected 0)')
print(f'std       = {deviation:.1f} degrees  (Gaussian approximation: {sigma_deg:.1f}, good only if eps >> kT)')
print(f'std-err   = {error:.2f} degrees  (std/sqrt(N) = {deviation/np.sqrt(num_spin):.2f})')

plt.hist(theta_w, bins=60, range=(-180, 180), density=True, color='lightgrey', label='spins')
plt.plot(grid, rho_w, 'k', label='Boltzmann')
plt.plot(grid, norm.pdf(grid, 0, sigma_deg), 'r--', label='Gaussian approximation')
plt.xlabel('angle (degrees)')
plt.ylabel('probability density (1/degree)')
plt.legend()
plt.show()
```

**The role of temperature.** The thermal kicks (energy $\sim k_BT$) erase the memory of the starting point: outcomes within a few $k_BT$ are all seen, outcomes many $k_BT$ higher almost never. Let's change $k_BT$ in units of $\varepsilon$: at high temperature the wheel does not feel the weight anymore (flat density, like the fair wheel), at low temperature it is always close to the minimum.

```
U_unit = 1 - np.cos(np.radians(grid))              # energy in units of eps

for kT_eps in [0.05, 0.2, 1, 5, 50]:
    w = np.exp(-U_unit/kT_eps)
    plt.plot(grid, w/(np.sum(w)*dtheta), label=f'kT/eps = {kT_eps}')
plt.xlabel('angle (degrees)')
plt.ylabel('probability density (1/degree)')
plt.legend()
plt.show()
```

**And back: from the histogram to the energy.** As for the die, the histogram gives us the energy: $U(\theta) = -k_BT\ln\rho(\theta) + \text{const}$. The constant is arbitrary (only energy *differences* matter), so we set the minimum to zero. Where are we able to reconstruct the energy and where not?

```
counts_w, edges = np.histogram(theta_w, bins=36, range=(-180, 180))
centres = 0.5*(edges[1:] + edges[:-1])
ok = counts_w > 0                                  # we can take the logarithm only where we have counts
U_est = -kT*np.log(counts_w[ok])
U_err = kT/np.sqrt(counts_w[ok])                   # error on ln(count) is about 1/sqrt(count)
U_est = U_est - U_est.min()                        # set the minimum to zero

plt.plot(grid, U/kT, 'k', label='exact')
plt.errorbar(centres[ok], U_est/kT, yerr=U_err/kT, fmt='o', capsize=3, label='from the histogram')
plt.xlabel('angle (degrees)')
plt.ylabel('energy U (kT)')
plt.legend()
plt.show()
print('bins without any spin:', np.sum(~ok), 'out of', len(counts_w))
```

The energy is well reconstructed where the wheel often goes (low energy) and it is missing, or has a large error, where it almost never goes (many $k_BT$ above the minimum): **this is the problem of sampling**, and it will come back in the next lectures.

**Cumulative distribution.** Another way of looking at the same data is the cumulative distribution function $F(\theta)=P(\theta'<\theta)$, that is the probability from $-\infty$ to $\theta$: it goes from 0 to 1 and it does not need any binning. The probability of an interval is $P(a<\theta<b)=F(b)-F(a)$.

```
X = np.sort(theta_w)
F = np.arange(1, len(X)+1)/len(X)

plt.plot(X, F, label='from the spins')
plt.plot(grid, np.cumsum(rho_w)*dtheta, 'k--', label='integral of the Boltzmann density')
plt.xlabel('angle (degrees)')
plt.ylabel('cumulative probability F')
plt.legend()
plt.show()
```

### 2D continuous data

Probability distributions can have more than one variable, for example the $(x,y)$ position of a molecule around a receptor. We use again a Gaussian, now in 2D. It is characterised by the average value of $x$ and $y$ and by the **covariance matrix**, that combines the squared standard deviations along $x$ and $y$ (diagonal) and their covariance (off-diagonal).

```
# this is the average value over X and Y
mean = [-2, 10]
# this is the covariance matrix: [[var(x), cov(x,y)], [cov(x,y), var(y)]]
cov = [[1, 3], [3, 20]]
num_sample = 10000
x, y = rng.multivariate_normal(mean, cov, size=num_sample).T
print(x.shape, y.shape)

plt.figure(figsize=(5, 5))
plt.plot(x, y, '.', markersize=2, alpha=0.3)       # small and transparent points, there are many
plt.axis('equal')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

With so many points the scatter plot is saturated, a **2D histogram** shows the density. As for the die, the number of bins matters: too few and we lose details, too many and every bin has few counts (sampling again).

```
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, nb in zip(axes, [8, 25, 100]):
    h = ax.hist2d(x, y, bins=(nb, nb), density=True, cmap='viridis')
    fig.colorbar(h[3], ax=ax, label='probability density')
    ax.set_title(f'{nb} x {nb} bins')
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.axis('equal')
plt.tight_layout()
plt.show()
```

**Projection (marginalisation).** To show a distribution in fewer dimensions we integrate over the other variable: $\rho(x)=\int\rho(x,y)\,dy$. With data this simply means making the histogram of $x$ alone, ignoring $y$ (and vice versa). The projections of a 2D Gaussian are 1D Gaussians with the average and standard deviation of the variable (the square root of the diagonal of the covariance).

```
fig, axes = plt.subplots(1, 2, figsize=(9, 3.5))
for ax, v, name, m, s in zip(axes, [x, y], ['x', 'y'], mean, np.sqrt(np.diag(cov))):
    ax.hist(v, bins=50, density=True, color='lightgrey', label='projection of the data')
    vg = np.linspace(v.min(), v.max(), 300)
    ax.plot(vg, norm.pdf(vg, m, s), 'k', label='Gaussian')
    ax.set_xlabel(name)
    ax.set_ylabel(f'density of {name}')
    ax.legend()
plt.tight_layout()
plt.show()

print(f'x: average = {np.average(x):.2f} +/- {sem(x):.2f},   std = {np.std(x):.2f}')
print(f'y: average = {np.average(y):.2f} +/- {sem(y):.2f},   std = {np.std(y):.2f}')
```

**Correlation.** If two variables are independent then $\rho(x,y)=\rho(x)\rho(y)$; if not they are correlated. The covariance matrix and the Pearson correlation coefficient $r=\mathrm{cov}(x,y)/(\sigma_x\sigma_y)$ (between $-1$ and $1$, zero for independent variables) quantify it. Here the expected value is $r=3/\sqrt{1\cdot 20}\simeq 0.67$. For comparison we also generate $x$ and $y$ with the same widths but independent.

```
# covariance matrix (x with x, x with y, y with x, y with y)
print('covariance matrix of the data:')
print(np.cov(x, y))
# correlation matrix (x with x, x with y, y with x, y with y)
print('correlation matrix of the data:')
print(np.corrcoef(x, y))

# same widths, but independent variables: no off-diagonal terms in the covariance
x_i, y_i = rng.multivariate_normal(mean, [[1, 0], [0, 20]], size=num_sample).T
print('correlation coefficient of the independent variables:', np.corrcoef(x_i, y_i)[0, 1])

fig, axes = plt.subplots(1, 2, figsize=(9, 4), sharex=True, sharey=True)
axes[0].plot(x, y, '.', markersize=2, alpha=0.3)
axes[0].set_title(f'correlated, r = {np.corrcoef(x, y)[0, 1]:.2f}')
axes[1].plot(x_i, y_i, '.', markersize=2, alpha=0.3)
axes[1].set_title(f'independent, r = {np.corrcoef(x_i, y_i)[0, 1]:.2f}')
for ax in axes:
    ax.set_xlabel('x')
axes[0].set_ylabel('y')
plt.tight_layout()
plt.show()
```
